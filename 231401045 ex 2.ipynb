```python
import pandas as pd

df = pd.read_excel(r"C:\Users\AI_LAB\Downloads\231401045\Teams.xlsx")
print(df.head())
```

```python
import sqlite3

# Create/connect to your database
conn = sqlite3.connect("teams.db")

# Store the DataFrame as a table named 'teams'
df.to_sql("teams", conn, if_exists="replace", index=False)

print("Database and table created successfully!")
```

```python
query = "SELECT * FROM teams"

result = pd.read_sql(query, conn)
print(result)
```
