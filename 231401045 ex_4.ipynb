```python
import pandas as pd

df = pd.read_csv(r"C:\Users\AI_LAB\Downloads\archive\Teams.csv")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_wins = df[df["W"] > 90]

print(high_wins)
```

```python
american_league = df[df["lgID"] == "AL"]

print(american_league)
```

```python
sorted_df = df.sort_values(
    by="W",
    ascending=False
)

print(sorted_df)
```

```python
print("Mean =",
      df["W"].mean())

print("Median =",
      df["W"].median())

print("Mode =")
print(df["W"].mode())

print("Standard Deviation =",
      df["W"].std())
```

```python
league_avg = df.groupby("lgID")["W"].mean()

print(league_avg)
```

```python
corr = df[[
    "W",
    "R"
]].corr()

print(corr)
```
