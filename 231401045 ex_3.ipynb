```python
import pandas as pd

df = pd.read_csv(r"C:\Users\AI_LAB\Downloads\archive\Teams.csv")

print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df["attendance"].fillna(df["attendance"].mean(), inplace=True)

df["ERA"].fillna(df["ERA"].mean(), inplace=True)

print(df)
```

```python
print(df.duplicated())

df.drop_duplicates(inplace=True)

print(df)
```

```python
Q1 = df['W'].quantile(0.25)
Q3 = df['W'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['W'] < lower) |
              (df['W'] > upper)]

print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['attendance']] = scaler.fit_transform(df[['attendance']])

print(df[['attendance']])
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df['teamID'] = encoder.fit_transform(df['teamID'])

print(df[['teamID']])
```

```python
import pandas as pd

team_encoded = pd.get_dummies(df['teamID'])

print(team_encoded)
```
