```python
import pandas as pd
df=pd.read_csv("Batting.csv")
print("First 5 records of Baseball dataset:")
display(df.head())
print("Dataset Shape:",df.shape)
print("Number of Columns:",len(df.columns))
```

```python
import matplotlib.pyplot as plt
season_runs=df.groupby("yearID")["R"].sum()
plt.figure(figsize=(10,5)); plt.plot(season_runs.index,season_runs.values,marker="o"); plt.title("Baseball Total Runs by Year"); plt.xlabel("Year"); plt.ylabel("Total Runs"); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
