```python
from pathlib import Path

def find_data(filename):
    candidates = [
        Path(filename),
        Path("data") / filename,
        Path("../data") / filename,
        Path("../../data") / filename,
        Path("/content") / filename,
    ]
    for p in candidates:
        if p.exists():
            return p
    raise FileNotFoundError(f"{filename} not found. Upload the Baseball Databank CSV or keep it in the repository data folder.")

import pandas as pd

df = pd.read_csv(find_data("Batting.csv"))

print("First 5 records of Baseball batting dataset:")
display(df.head())

print("Dataset Shape:", df.shape)
print("Number of Columns:", len(df.columns))
```

```python
import matplotlib.pyplot as plt

season_runs = df.groupby("yearID")["R"].sum()

plt.figure(figsize=(10, 5))
plt.plot(season_runs.index, season_runs.values, marker="o")
plt.title("Baseball Total Runs by Year")
plt.xlabel("Year")
plt.ylabel("Total Runs")
plt.xticks(season_runs.index[::max(1, len(season_runs)//15)], rotation=45)
plt.tight_layout()
plt.show()
```
