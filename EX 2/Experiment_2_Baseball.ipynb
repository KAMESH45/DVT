## DATA IMPORT AND EXPORT – BASEBALL DATABANK
**Aim:** Import, inspect, export and transfer Baseball data using CSV, Excel, JSON and SQLite.

```python
import pandas as pd
import sqlite3
batting=pd.read_csv("Batting.csv")
print("Baseball CSV file imported successfully.")
print("Rows:",len(batting)); print("Columns:",len(batting.columns)); display(batting.head())
```

```python
batting.info(); display(batting.describe(include="all").T)
```

```python
sample=batting.head(20); sample.to_excel("baseball_sample.xlsx",index=False); excel=pd.read_excel("baseball_sample.xlsx"); print("Excel file imported successfully."); display(excel.head())
```

```python
conn=sqlite3.connect("baseball_data.db"); sample.to_sql("batting_records",conn,if_exists="replace",index=False); sql=pd.read_sql("SELECT * FROM batting_records LIMIT 5",conn); print("SQL data imported successfully."); display(sql); conn.close()
```

```python
print("Web import syntax demonstrated using pd.read_csv(url).")
print("Example: pd.read_csv(url)")
```

```python
sample.to_csv("baseball_export.csv",index=False); print("Baseball CSV file exported successfully."); print("Exported rows:",len(sample))
sample.to_excel("baseball_export.xlsx",index=False); print("Baseball Excel file exported successfully."); print("Exported rows:",len(sample))
sample.to_json("baseball_export.json",orient="records",indent=2); print("Baseball JSON file exported successfully."); print("Exported records:",len(sample))
```

```python
conn=sqlite3.connect("baseball_export.db"); sample.to_sql("baseball_data",conn,if_exists="replace",index=False); count=pd.read_sql("SELECT COUNT(*) AS row_count FROM baseball_data",conn); print("Baseball data exported to SQLite successfully."); display(count); conn.close()
```
