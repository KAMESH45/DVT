# EX 2 – Baseball Databank

Same import/export workflow as the friend's EX 2, using Baseball Databank Batting.csv.

```python
import pandas as pd
import sqlite3
```

```python
from pathlib import Path

def find_data(filename):
    candidates = [
        Path(filename),
        Path("data") / filename,
        Path("../data") / filename,
        Path("../../data") / filename,
        Path("/content") / filename,
    ]
    for p in candidates:
        if p.exists():
            return p
    raise FileNotFoundError(f"{filename} not found. Upload the Baseball Databank CSV or keep it in the repository data folder.")

batting = pd.read_csv(find_data("Batting.csv"))
print("Baseball CSV file imported successfully.")
print("Rows:", len(batting))
print("Columns:", len(batting.columns))
```

```python
batting_sample = batting.head(20)
batting_sample.to_excel("baseball_sample.xlsx", index=False)

batting_excel = pd.read_excel("baseball_sample.xlsx")
print("Excel file imported successfully.")
display(batting_excel.head())
```

```python
conn = sqlite3.connect("baseball_data.db")
batting_sample.to_sql("batting_records", conn, if_exists="replace", index=False)

query = "SELECT * FROM batting_records LIMIT 5"
batting_sql = pd.read_sql(query, conn)

print("SQL data imported successfully.")
display(batting_sql)
conn.close()
```

```python
print("Web import syntax demonstrated using pd.read_csv(url).")
print("Example: pd.read_csv(url) where url points to a CSV resource.")
```

```python
batting_sample.to_csv("baseball_export.csv", index=False)
print("Baseball CSV file exported successfully.")
print("Exported rows:", len(batting_sample))
```

```python
batting_sample.to_excel("baseball_export.xlsx", index=False)
print("Baseball Excel file exported successfully.")
print("Exported rows:", len(batting_sample))
```

```python
batting_sample.to_json("baseball_export.json", orient="records", indent=2)
print("Baseball JSON file exported successfully.")
print("Exported records:", len(batting_sample))
```

```python
conn = sqlite3.connect("baseball_export.db")
batting_sample.to_sql("baseball_data", conn, if_exists="replace", index=False)

count = pd.read_sql("SELECT COUNT(*) AS row_count FROM baseball_data", conn)
print("Baseball data exported to SQLite successfully.")
display(count)
conn.close()
```
